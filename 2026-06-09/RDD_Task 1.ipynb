```python
freq_rdd = rdd.map(lambda x: (x, 1)) \
              .reduceByKey(lambda a, b: a + b)
```

```python
print(
    freq_rdd.sortBy(lambda x: x[1], ascending=False)
            .first()
)
```

```python
print(
    freq_rdd.sortBy(lambda x: x[1])
            .first()
)
```

```python
login_count = rdd.filter(lambda x: x == "LOGIN").count()
total = rdd.count()

print((login_count / total) * 100)
```

```python
video_count = rdd.filter(lambda x: x == "VIDEO_PLAY").count()
total = rdd.count()

print((video_count / total) * 100)
```
