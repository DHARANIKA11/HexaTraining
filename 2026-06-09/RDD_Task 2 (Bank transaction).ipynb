```python
print("========== BANK DASHBOARD ==========")

print("Total Transactions :", rdd.count())

print("Unique Transaction Types :", rdd.distinct().count())

print("\nTransaction Frequencies")

freq = rdd.map(lambda x: (x, 1)) \
          .reduceByKey(lambda a, b: a + b)

for i in freq.collect():
    print(i)

print("\nTotal Money Processed :", amount_rdd.sum())

print("Average Money Processed :", amount_rdd.mean())

print("Maximum Amount :", amount_rdd.max())

print("Minimum Amount :", amount_rdd.min())

print("\nHigh Value Transactions (>20000)")
print(amount_rdd.filter(lambda x: x > 20000).collect())

print("\nLow Value Transactions (<1000)")
print(amount_rdd.filter(lambda x: x < 1000).collect())

print("===================================")
```
