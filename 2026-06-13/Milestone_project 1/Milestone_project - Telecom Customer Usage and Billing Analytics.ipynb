```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("TelecomAnalytics") \
    .getOrCreate()
```

```python
customers_df = spark.read.csv(
    "customers.csv",
    header=True,
    inferSchema=True
)

customers_df.show()
```

```python
usage_df = spark.read.csv(
    "usage.csv",
    header=True,
    inferSchema=True
)

usage_df.show()
```

```python
payments_df = spark.read.csv(
    "payments.csv",
    header=True,
    inferSchema=True
)

payments_df.show()
```

```python
plans_df = spark.read.option(
    "multiline",
    "true"
).json("plans.json")

plans_df.show(truncate=False)
```

```python
usage_df = usage_df.fillna({
    "data_used_gb": 0
})

usage_df.show()
```

```python
payments_df = payments_df.fillna({
    "amount_paid": 0
})

payments_df.show()
```

```python
payments_df = payments_df.fillna({
    "payment_mode": "Not Provided"
})

payments_df.show()
```

```python
from pyspark.sql.functions import when, col

customers_df = customers_df.withColumn(
    "plan_id",
    when(
        (col("plan_id").isNull()) | (col("plan_id") == ""),
        "UNKNOWN"
    ).otherwise(col("plan_id"))
)

customers_df.show()
```

```python
customers_df = customers_df.withColumn(
    "data_quality_status",
    when(col("plan_id") == "UNKNOWN", "Incomplete")
    .otherwise("Complete")
)

customers_df.show()
```

```python
customers_df.write.mode("overwrite").parquet("silver/customers")

usage_df.write.mode("overwrite").parquet("silver/usage")

payments_df.write.mode("overwrite").parquet("silver/payments")

plans_df.write.mode("overwrite").parquet("silver/plans")
```

```python
from pyspark.sql.functions import when, col

customer_usage_billing_df = customer_usage_billing_df.withColumn(
    "usage_category",
    when(col("data_used_gb") >= 70, "Heavy User")
    .when(col("data_used_gb") >= 30, "Medium User")
    .otherwise("Low User")
)

customer_usage_billing_df.select(
    "customer_id",
    "data_used_gb",
    "usage_category"
).show()
```

```python
customer_usage_billing_df = customer_usage_billing_df.withColumn(
    "payment_category",
    when(col("amount_paid") >= 1000, "High Payment")
    .when(col("amount_paid") >= 500, "Medium Payment")
    .otherwise("Low Payment")
)

customer_usage_billing_df.select(
    "customer_id",
    "amount_paid",
    "payment_category"
).show()
```

```python
customer_usage_billing_df = customer_usage_billing_df.withColumn(
    "churn_risk",
    when(
        (col("status") == "Inactive") |
        (col("payment_status") != "Success"),
        "High Risk"
    )
    .when(
        col("data_used_gb") < 15,
        "Medium Risk"
    )
    .otherwise("Low Risk")
)

customer_usage_billing_df.select(
    "customer_id",
    "status",
    "payment_status",
    "data_used_gb",
    "churn_risk"
).show()
```

```python
customer_usage_billing_df = customer_usage_billing_df.withColumn(
    "over_usage_gb",
    col("data_used_gb") - col("data_limit_gb")
)

customer_usage_billing_df.select(
    "customer_id",
    "data_used_gb",
    "data_limit_gb",
    "over_usage_gb"
).show()
```

```python
customer_usage_billing_df = customer_usage_billing_df.withColumn(
    "over_usage_flag",
    when(col("over_usage_gb") > 0, "Yes")
    .otherwise("No")
)

customer_usage_billing_df.select(
    "customer_id",
    "over_usage_gb",
    "over_usage_flag"
).show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import rank

data_usage_window = Window.orderBy(col("data_used_gb").desc())

customer_usage_billing_df = customer_usage_billing_df.withColumn(
    "data_usage_rank",
    rank().over(data_usage_window)
)

customer_usage_billing_df.select(
    "customer_id",
    "customer_name",
    "data_used_gb",
    "data_usage_rank"
).show()
```

```python
payment_window = Window.orderBy(col("amount_paid").desc())

customer_usage_billing_df = customer_usage_billing_df.withColumn(
    "payment_rank",
    rank().over(payment_window)
)

customer_usage_billing_df.select(
    "customer_id",
    "customer_name",
    "amount_paid",
    "payment_rank"
).show()
```

```python
from pyspark.sql.functions import row_number

city_window = Window.partitionBy("city").orderBy(col("amount_paid").desc())

top_customer_city_df = customer_usage_billing_df.withColumn(
    "city_rank",
    row_number().over(city_window)
)

top_customer_city_df.filter(
    col("city_rank") == 1
).select(
    "city",
    "customer_id",
    "customer_name",
    "amount_paid"
).show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, col

plan_window = Window.partitionBy("plan_name").orderBy(col("amount_paid").desc())

top_customer_plan_df = customer_usage_billing_df.withColumn(
    "plan_rank",
    row_number().over(plan_window)
)

top_customer_plan_df.filter(
    col("plan_rank") == 1
).select(
    "plan_name",
    "customer_id",
    "customer_name",
    "amount_paid"
).show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import sum

month_window = Window.orderBy("bill_month") \
                     .rowsBetween(Window.unboundedPreceding, Window.currentRow)

monthly_revenue_df = customer_usage_billing_df.groupBy("bill_month") \
    .agg(
        sum("amount_paid").alias("monthly_revenue")
    ) \
    .orderBy("bill_month")

monthly_revenue_df = monthly_revenue_df.withColumn(
    "running_total_revenue",
    sum("monthly_revenue").over(month_window)
)

monthly_revenue_df.show()
```

```python
from pyspark.sql.functions import lag

usage_window = Window.partitionBy("customer_id").orderBy("usage_month")

usage_lag_df = usage_df.withColumn(
    "previous_month_usage",
    lag("data_used_gb", 1).over(usage_window)
)

usage_lag_df.show()
```

```python
from pyspark.sql.functions import lead

usage_lead_df = usage_df.withColumn(
    "next_month_usage",
    lead("data_used_gb", 1).over(usage_window)
)

usage_lead_df.show()
```

```python
from pyspark.sql.functions import col

usage_increase_df = usage_lag_df.filter(
    col("data_used_gb") > col("previous_month_usage")
)

usage_increase_df.select(
    "customer_id",
    "usage_month",
    "previous_month_usage",
    "data_used_gb"
).show()
```
