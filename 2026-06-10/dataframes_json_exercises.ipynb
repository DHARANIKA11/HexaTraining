```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Working with JSON") \
    .getOrCreate()
```

```python
hotels_df = spark.read.option(
    "multiline",
    "true"
).json("Hotels.json")
```

```python
hotels_df.show(truncate=False)
```

```python
hotels_df.show(truncate=False)
```

```python
hotels_df.select("hotel_name", "city", "rating").show()
```

```python
hotels_df.filter(hotels_df.category == "Luxury").show()
```

```python
hotels_df.filter(hotels_df.rating > 4.5).show()
```

```python
hotels_df.filter(hotels_df.rooms_available > 15).show()
```

```python
hotels_df.filter(hotels_df.price_per_night > 10000).show()
```

```python
hotels_df.filter(
    (hotels_df.city == "Dubai") |
    (hotels_df.city == "London")
).show()
```

```python
from pyspark.sql.functions import col

hotels_df.filter(col("contact.phone").isNull()).show(truncate=False)
```

```python
from pyspark.sql.functions import col

hotels_df.filter(col("contact.email").isNull()).show(truncate=False)
```

```python
hotels_df.select(
    "hotel_name",
    "contact.phone",
    "contact.email"
).show(truncate=False)
```

```python
from pyspark.sql.functions import array_contains

hotels_df.filter(
    array_contains("amenities", "wifi")
).show(truncate=False)
```

```python
from pyspark.sql.functions import array_contains

hotels_df.filter(
    array_contains("amenities", "spa")
).show(truncate=False)
```

```python
hotels_df.groupBy("city").count().show()
```

```python
hotels_df.groupBy("category").count().show()
```

```python
from pyspark.sql.functions import avg

hotels_df.groupBy("category") \
    .agg(avg("rating").alias("Average_Rating")) \
    .show()
```

```python
from pyspark.sql.functions import avg

hotels_df.groupBy("city") \
    .agg(avg("price_per_night").alias("Average_Price")) \
    .show()
```

```python
from pyspark.sql.functions import max

hotels_df.agg(
    max("price_per_night").alias("Highest_Price")
).show()
```

```python
from pyspark.sql.functions import col

hotels_df = hotels_df.withColumn(
    "total_potential_revenue",
    col("rooms_available") * col("price_per_night")
)

hotels_df.show(truncate=False)
```

```python
hotels_df.orderBy(
    hotels_df.rating.desc()
).show(truncate=False)
```

```python
flat_df = hotels_df.select(
    "hotel_id",
    "hotel_name",
    "city",
    "category",
    "rating",
    "rooms_available",
    "price_per_night",
    "amenities",
    "contact.phone",
    "contact.email"
)

flat_df.write.mode("overwrite").parquet("hotels_flattened.parquet")
```
