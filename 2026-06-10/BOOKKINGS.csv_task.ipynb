```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Booking Data Analysis") \
    .getOrCreate()
```

```python
df = spark.read.csv(
    "BOOKKINGS.csv",
    header=True,
    inferSchema=True
)
```

```python
from pyspark.sql.functions import col, sum

df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
]).show()
```

```python
clean_df = df.na.drop()

clean_df.show()
```

```python
df.na.drop(subset=["booking_amount"]).show()
```

```python
df.na.drop(subset=["customer_name", "service_type", "booking_amount"]).show()
```

```python
df = df.fillna({"city": "Unknown"})
df.show()
```

```python
df = df.fillna({"provider": "Not Available"})
df.show()
```

```python
df = df.fillna({"payment_mode": "Not Provided"})
df.show()
```

```python
df = df.fillna({"booking_status": "Unknown"})
df.show()
```

```python
df = df.fillna({"booking_amount": 0})
df.show()
```

```python
from pyspark.sql.functions import when, col

df = df.withColumn(
    "data_quality_status",
    when(
        col("city").isNull() |
        col("service_type").isNull() |
        col("provider").isNull() |
        col("booking_amount").isNull() |
        col("booking_status").isNull() |
        col("payment_mode").isNull(),
        "Incomplete"
    ).otherwise("Complete")
)

df.show()
```

```python
df.groupBy("data_quality_status").count().show()
```

```python
df.filter(df.data_quality_status == "Complete").show()
```

```python
df.filter(df.data_quality_status == "Incomplete").show()
```

```python
from pyspark.sql.functions import col

df = df.withColumn("tax", col("booking_amount") * 0.05)

df.show()
```

```python
from pyspark.sql.functions import col

df = df.withColumn("tax", col("booking_amount") * 0.05)

df.show()
```

```python
from pyspark.sql.functions import col

df = df.withColumn("final_amount", col("booking_amount") + col("tax"))

df.show()
```

```python
from pyspark.sql.functions import sum

df.filter(df.booking_status == "Confirmed") \
  .agg(sum("booking_amount").alias("Total_Revenue")) \
  .show()
```

```python
df.fillna({"service_type": "Unknown"}) \
  .groupBy("service_type") \
  .count() \
  .show()
```

```python
df.fillna({"city": "Unknown"}) \
  .groupBy("city") \
  .count() \
  .show()
```

```python
from pyspark.sql.functions import avg

df.fillna({"booking_amount": 0}) \
  .agg(avg("booking_amount").alias("Average_Amount")) \
  .show()
```

```python
from pyspark.sql.functions import avg

df.na.drop(subset=["booking_amount"]) \
  .agg(avg("booking_amount").alias("Average_Amount")) \
  .show()
```

```python
from pyspark.sql.functions import avg

avg_fill = df.fillna({"booking_amount": 0}) \
             .agg(avg("booking_amount").alias("Avg_With_Zero"))

avg_drop = df.na.drop(subset=["booking_amount"]) \
             .agg(avg("booking_amount").alias("Avg_Without_Null"))

avg_fill.show()
avg_drop.show()
```
