```python
from pyspark.sql import functions as F
from pyspark.sql.functions import col, when

sales_df = df.withColumn(
    "total_amount", col("quantity") * col("amount")
).withColumn(
    "discount", col("total_amount") * 0.10
).withColumn(
    "final_amount", col("total_amount") - col("discount")
)
```

```python
sales_df.orderBy(col("total_amount").desc()).show(3)
```

```python
sales_df.orderBy(col("total_amount")).show(1)
```

```python
sales_df.filter(col("product").startswith("M")).show()
```

```python
sales_df.filter(
    F.lower(col("customer_name")).contains("a")
).show()
```

```python
sales_df.filter(
    F.lower(col("product")).contains("o")
).show()
```

```python
sales_df = sales_df.withColumn(
    "order_value_category",
    when(col("total_amount") >= 50000, "High")
    .when(col("total_amount") >= 20000, "Medium")
    .otherwise("Low")
)

sales_df.show()
```

```python
sales_df.groupBy("order_value_category") \
        .count() \
        .show()
```

```python
sales_df.filter(col("status") == "Delivered") \
        .select(
            "order_id",
            "customer_name",
            "product",
            "final_amount"
        ) \
        .show()
```

```python
sales_df.filter(col("status") == "Delivered") \
        .groupBy("city") \
        .agg(
            F.sum("final_amount").alias("delivered_revenue")
        ) \
        .show()
```

```python
final_report = sales_df.select(
    "order_id",
    "customer_name",
    "city",
    "category",
    "product",
    "quantity",
    col("amount").alias("price"),
    "total_amount",
    "discount",
    "final_amount",
    col("status").alias("order_status")
)

final_report.show()
```
